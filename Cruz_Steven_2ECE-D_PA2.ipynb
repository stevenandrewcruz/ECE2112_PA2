EXPERIMENT 2: NUMERICAL PYTHON (NUMPY)

```python
import numpy as np
```

A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5)) 
X
```

```python
mean = np.mean(X)
mean
```

```python
std = np.std(X)
std
```

```python
Difference = X - mean
Difference
```

```python
X_normalized = Difference/std
X_normalized
```

```python
mean_X_nomralized = np.mean(X_normalized)
mean_X_nomralized
```

```python
print("X array:\n", X)
```

```python
print("mean:\n", mean)
```

```python
print("Standard Deviation:\n", std)
```

```python
print("Z=", X_normalized)
```

```python
print("Z mean:\n", mean_X_nomralized)
```

```python
np.save('X_normalized.npy', X_normalized)
```

B. CUBES DIVISIBLE BY 4 PROBLEM

```python
c1=  (np.arange(1,101)).reshape(10,10)
c1
```

```python
C = np.power= c1**3
C
```

```python
div_by_4= C[C % 4 ==0]
div_by_4
```

```python
print('C:',C)
```

```python
print('Cubed values divisible by 4:\n ',div_by_4)
```

```python
np.save('div_by_4.npy', div_by_4)
```

C. ABOVE-MEAN SQUARES PROBLEM

```python
S1 = (np.arange(1,37)).reshape(6,6)
S1
```

```python
S = np.power = S1**2
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S>S_mean]
above_mean 
```

```python
print('S:',S)
```

```python
print('Mean:\n',S_mean)
```

```python
print('Above the mean:\n',above_mean)
```

```python
np.save('above_mean.npy', above_mean)
```
